# Visualizations

Dataset: The <a href="https://archive.ics.uci.edu/ml/datasets/wine+quality">Wine Quality Data Set</a> from the UCI Machine Learning Repository. 

The two datasets are related to red and white variants of the Portuguese "Vinho Verde" wine.

We will be using seaborn for most plots

```python
%matplotlib inline
import numpy as np
from pandas import Series, DataFrame
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

## Preprocessing

```python
red_wine = pd.read_csv('data/winequality-red.csv', sep=';')
white_wine = pd.read_csv('data/winequality-white.csv', sep=';')

# store wine type as an attribute
red_wine['wine_type'] = 'red'   
white_wine['wine_type'] = 'white'

# bucket wine quality scores into qualitative quality labels (using pd.Categorical to specify the order of the labels)
red_wine['quality_label'] = red_wine['quality'].apply(lambda value: 'low' 
                                                          if value <= 5 else 'medium' 
                                                              if value <= 7 else 'high')
red_wine['quality_label'] = pd.Categorical(red_wine['quality_label'], 
                                           categories=['low', 'medium', 'high'])
white_wine['quality_label'] = white_wine['quality'].apply(lambda value: 'low' 
                                                              if value <= 5 else 'medium' 
                                                                  if value <= 7 else 'high')
white_wine['quality_label'] = pd.Categorical(white_wine['quality_label'], 
                                             categories=['low', 'medium', 'high'])

# merge red and white wine datasets
wines = pd.concat([red_wine, white_wine])

# re-shuffle records just to randomize data points
wines = wines.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
wines.head()
```

We have several numeric and categorical attributes for wine samples. Each observation belongs to a red or white wine sample and the attributes are specific attributes or properties measured and obtained from physicochemical tests.

```python
subset_attributes = ['residual sugar', 'total sulfur dioxide', 'sulphates', 
                     'alcohol', 'volatile acidity', 'quality']
rs = round(red_wine[subset_attributes].describe(),2)
ws = round(white_wine[subset_attributes].describe(),2)

pd.concat([rs, ws], axis=1, keys=['Red Wine Statistics', 'White Wine Statistics'])
```

## Univariate

```python
# histograms using pandas

wines.hist(bins=15, color='steelblue', edgecolor='black', linewidth=1.0,
           xlabelsize=8, ylabelsize=8, grid=False)    
plt.tight_layout(rect=(0, 0, 1.2, 1.2)) 
```

```python
# histogram
_=sns.histplot(data=wines.sulphates, bins=20)
```

```python
# density plot
sns.kdeplot(data=wines.sulphates, fill=True, color='steelblue');
```

```python
# histogram incl density plot
sns.histplot(data=wines.sulphates, bins=20, kde=True);
```

```python
# boxplot
sns.boxplot(data = wines.sulphates);
```

```python
# violin plot
sns.violinplot(data=wines.sulphates);
```

```python
# frequency plot 
sns.countplot(data=wines, x="quality_label");
```

```python
# frequency plot - rotated by 90 degrees
sns.countplot(data=wines, y="quality_label");
```

## Bivariate

```python
# Scatter plot
sns.scatterplot(data=wines, x="fixed acidity", y="sulphates");
```

```python
# Joint plot
sns.jointplot(data=wines, x="fixed acidity", y="sulphates");
```

```python
# multi-bar plot
cp = sns.countplot(x="quality", hue="wine_type", data=wines, 
                   palette={"red": "#FF9999", "white": "#FFE888"});
```

```python
# Multi Box Plots
f, ax = plt.subplots(1, 1, figsize=(12, 4))
f.suptitle('Wine Quality - Alcohol Content', fontsize=14)

sns.boxplot(x="quality", y="alcohol", data=wines,  ax=ax)
ax.set_xlabel("Wine Quality",size = 12,alpha=0.8)
ax.set_ylabel("Wine Alcohol %",size = 12,alpha=0.8);
```

```python
# Multi Violin Plots
f, ax = plt.subplots(1, 1, figsize=(12, 4))
f.suptitle('Wine Quality - Alcohol Content', fontsize=14)

sns.violinplot(x="quality", y="alcohol", data=wines,  ax=ax)
ax.set_xlabel("Wine Quality",size = 12,alpha=0.8)
ax.set_ylabel("Wine Alcohol %",size = 12,alpha=0.8);
```

## Bivariate for multiple attributes

```python
# Pair-wise Correlation Matrix Heatmap
f, ax = plt.subplots(figsize=(10, 6))
corr = wines.corr(numeric_only=True)
hm = sns.heatmap(round(corr,2), annot=True, ax=ax, cmap="coolwarm",fmt='.2f',
                 linewidths=.05)
f.subplots_adjust(top=0.93)
t= f.suptitle('Wine Attributes Correlation Heatmap', fontsize=14)
```

```python
# Pair-wise Scatter Plots
cols = ['density', 'residual sugar', 'total sulfur dioxide', 'fixed acidity']
sns.pairplot(wines[cols]);
```

```python
# Pair-wise Scatter Plots - a little nicer
cols = ['density', 'residual sugar', 'total sulfur dioxide', 'fixed acidity']
pp = sns.pairplot(wines[cols], height=2, aspect=1.5, # set height of each facet to 2 inches and width to 2*1.5 inches
                  plot_kws=dict(edgecolor="black", linewidth=0.5), # change edges of points to thin black
                  diag_kind="kde", diag_kws=dict(fill=True)) # set diagonal to density charts, kws=keyword arguments

fig = pp.fig 
fig.subplots_adjust(top=0.93, wspace=0.3) # increase top space and white spaces between the plots
fig.suptitle('Wine Attributes Pairwise Plots', fontsize=14); # add title to figure
```

```python
# Strip Plot
wines_sample = wines.sample(frac=0.2, random_state=42)
plt.figure(figsize=(16,6))
sns.stripplot(x=wines_sample['quality'], y=wines_sample['residual sugar'], size=3);
```

```python
# Swarm Plot
wines_sample = wines.sample(frac=0.2, random_state=42)
plt.figure(figsize=(16,6))
sns.swarmplot(x=wines_sample['quality'], y=wines_sample['residual sugar'],  size=2);
```

## Multivariate

```python
# Scatter Plot with Hue 
cols = ['density', 'residual sugar', 'total sulfur dioxide', 'fixed acidity', 'wine_type']
pp = sns.pairplot(wines[cols], hue='wine_type', height=1.8, aspect=2, 
                  palette={"red": "#FF9999", "white": "#FFE888"},
                  plot_kws=dict(edgecolor="black", linewidth=0.5))
fig = pp.fig 
fig.subplots_adjust(top=0.93, wspace=0.3)
fig.suptitle('Wine Attributes Pairwise Plots with Wine Type as Hue', fontsize=14);
```

```python
# Scatter Plot with Hue
cols = ['density', 'residual sugar', 'total sulfur dioxide', 'fixed acidity', 'quality_label']
pp = sns.pairplot(wines[cols], hue='quality_label', height=1.8, aspect=2, 
                  #palette={"low": "#FF3333", "medium": "#33FF33", "high": "#3333FF"}, # either a dictionary
                  palette = sns.color_palette("pastel")[:3], # or a predefined palette
                  plot_kws=dict(edgecolor="black", linewidth=0.5))
fig = pp.fig 
fig.subplots_adjust(top=0.93, wspace=0.3)
fig.suptitle('Wine Attributes Pairwise Plots with Wine Quality Label as Hue', fontsize=14);
```

```python
# Multi Violinplot with Hue for Binary Attribute
sns.violinplot(data=wines, x="quality", y="volatile acidity", hue="wine_type", split=True, inner=None, palette={"red": "#FF9999", "white": "#FFE888"});
```

```python
# Swarm Plot with Hue as Third Dimension
wines_sample = wines.sample(frac=0.1, random_state=42)
plt.figure(figsize=(16,5))
sns.swarmplot(x=wines_sample['quality'], y=wines_sample['residual sugar'], hue=wines_sample['wine_type'],  size=3, dodge=True);
```

```python
# Swarm Plot with Hue as Third Dimension
wines_sample = wines.sample(frac=0.1, random_state=42)
plt.figure(figsize=(16,5))
sns.swarmplot(x=wines_sample['wine_type'], y=wines_sample['residual sugar'], hue=wines_sample['quality'],  size=3, dodge=True, palette='viridis');
```

```python
# Scatterplot with Hue as Third Dimension
wines_sample = wines.sample(frac=0.2, random_state=42)
sns.scatterplot(data=wines_sample, x="fixed acidity", y="sulphates", hue='quality_label', palette='flare');
```

```python
# Scatterplot with Size as Third Dimension(Bubble Chart)
wines_sample = wines.sample(frac=0.2, random_state=42)
sns.scatterplot(data=wines_sample, x="fixed acidity", y="sulphates", size='quality_label', size_order=['high', 'medium', 'low'], alpha=0.5);
```
